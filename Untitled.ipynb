```python
# for invalid values in some columns

# 1. invalid values -> NaN
for c in ["Physical Activity", "Diet Score", "Stress Level", "Triglycerides","Oxygen Saturation"]:
    df.loc[df[c] < 0, c] = np.nan
df["Oxygen Saturation"] = df["Oxygen Saturation"].clip(upper=100)

# 2. categorical gaps
df["Gender"] = df["Gender"].fillna("Unknown")
df["Medical Condition"] = df["Medical Condition"].fillna("Unknown")
```

```python
#imputation 
import pandas as pd
import numpy as np

df = pd.read_csv("hospital_dataset.csv")

cols = ["Age", "Blood Pressure", "Glucose","Oxygen Saturation"]

# 1. before: how many missing + summary stats
print("Missing before:\n", df[cols].isna().sum())
before = df[cols].describe().loc[["mean", "50%", "std"]]

# 2. (optional) flag which rows were imputed
for c in cols:
    df[f"{c}_was_missing"] = df[c].isna().astype(int)

# 3. impute with median
for c in cols:
    median_val = df[c].median()          # median ignores NaN automatically
    df[c] = df[c].fillna(median_val)
    print(f"{c}: filled with median = {median_val}")

# 4. Age should be a whole number
df["Age"] = df["Age"].round().astype(int)

# 5. after: confirm nothing is missing + compare stats
print("\nMissing after:\n", df[cols].isna().sum())
after = df[cols].describe().loc[["mean", "50%", "std"]]

print("\nBefore:\n", before.round(2))
print("\nAfter:\n", after.round(2))
```

```python
#for outliers
import pandas as pd, numpy as np

def iqr_bounds(s, k=1.5):
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    return q1 - k * iqr, q3 + k * iqr

cols = ["Age", "Glucose", "Blood Pressure", "BMI", "Oxygen Saturation",
        "LengthOfStay", "Cholesterol", "Triglycerides", "HbA1c", "Sleep Hours"]

summary = []
for c in cols:
    lo, hi = iqr_bounds(df[c].dropna())
    n = ((df[c] < lo) | (df[c] > hi)).sum()
    summary.append([c, round(lo, 1), round(hi, 1), n, round(100 * n / df[c].notna().sum(), 1)])

print(pd.DataFrame(summary, columns=["column", "lower", "upper", "n_outliers", "pct"]))


#to represent the outliers using box plot

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6, 4))
sns.boxplot(y=df["Glucose"], color="skyblue",
            flierprops=dict(marker="o", markersize=3, markerfacecolor="red",
                            markeredgecolor="none", alpha=0.4))
plt.title("Glucose: outliers shown as red dots")
plt.ylabel("Glucose (mg/dL)")
plt.show()


cols = ["Age", "Glucose", "Blood Pressure", "BMI",
        "LengthOfStay", "HbA1c", "Cholesterol", "Triglycerides"]

fig, axes = plt.subplots(2, 4, figsize=(16, 7))
for ax, c in zip(axes.flat, cols):
    sns.boxplot(y=df[c], ax=ax, color="lightsteelblue",
                flierprops=dict(markersize=2, markerfacecolor="red",
                                markeredgecolor="none", alpha=0.4))
    ax.set_title(c)
    ax.set_ylabel("")
plt.suptitle("Boxplots for outlier detection", fontsize=14)
plt.tight_layout()
plt.show()
```
